### 1.The historical average score in a certification exam is 65. A training institute introduces a new course. A random sample of 50 students trained under the new course has a mean score of 68 and a standard deviation of 10. At the 5% level of significance, test whether the new course improves performance.

```python
from scipy import stats
import numpy as np
```

```python
observations_50_students =np.random.normal(loc = 68,scale =10,size=50).astype(int)
observations_50_students
```

```python
t_stats,p_value =stats.ttest_1samp(a=observations_50_students,popmean=65,alternative='greater')
t_stats,p_value
```

```python
if p_value<0.05:
    print("We reject HO")
else:
    print("We do not reject HO")
```

### 5. Two teaching methods are used to train employees. A sample of 30 employees trained using Method A has a mean score of 75 with a standard deviation of 12. A sample of 28 employees trained using Method B has a mean score of 70 with a standard deviation of 10. At the 5% level of significance, test whether there is a difference in average scores between the two methods.



```python
method_A = np.random.normal(loc = 75,scale =12.0,size=30).astype(int)
method_B = np.random.normal(loc = 70,scale =10.0,size=28).astype(int)
t_stats,p_value =stats.ttest_ind(a=method_A,b=method_B,alternative='two-sided')
if p_value<0.05:
    print("We reject HO")
else:
    print("We do not reject HO")
t_stats,p_value
```

### 9. A company wants to compare the average productivity of employees trained using three different training programs. Program A has 25 employees with a mean productivity score of 80 and a standard deviation of 8. Program B has 30 employees with a mean of 75 and a standard deviation of 7. Program C has 28 employees with a mean of 82 and a standard deviation of 6. At the 5% level of significance, test whether there is a difference in mean productivity across the programs.

```python
program_A = np.random.normal(loc = 80,scale =8.0,size=25).astype(int)
program_B = np.random.normal(loc = 75,scale =7.0,size=30).astype(int)
program_C = np.random.normal(loc = 82,scale =6.0,size=28).astype(int)

f_stats,p_value =stats.f_oneway(program_A,program_B,program_C)
if p_value<0.05:
    print("We reject HO")
else:
    print("We do not reject HO")
f_stats,p_value
```

### 10. A company wants to test whether gender and product preference are independent. The survey results are as follows: Male: 60 prefer Product A, 40 prefer Product B,Female: 50 prefer Product A, 50 prefer Product B,At the 5% level of significance, test for independence.


```python
observations = np.array([[60,40],[50,50]])
chi_stats,pvalue,dof,expected=stats.chi2_contingency(observations)
if p_value<0.05:
    print("We reject HO")
else:
    print("We do not reject HO")
chi_stats,p_value
```

### 12.	A dice manufacturer claims that a die is fair. The die is rolled 120 times, and the observed frequencies are:
●	Face 1: 18
●	Face 2: 22
●	Face 3: 20
●	Face 4: 25
●	Face 5: 17
●	Face 6: 18
At the 5% level of significance, test whether the die is fair.



```python
observations_dies = np.array([18,22,20,25,17,18])
expected = np.array([20,20,20,20,20,20])

chi_stats,pvalue=stats.chisquare(f_obs=observations_dies ,f_exp=expected)
if p_value<0.05:
    print("We reject HO")
else:
    print("We do not reject HO")
chi_stats,p_value
```
